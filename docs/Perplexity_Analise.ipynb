# Investimentos e Ações em Segurança Pública em Minas Gerais: Comparação 2024 vs 2025

## Resumo Executivo

Minas Gerais mantém sua posição como referência nacional em segurança pública, com investimentos crescentes e ações estratégicas coordenadas. O estado aumentou o orçamento total destinado à segurança pública em **5,1%** entre 2024 e 2025, chegando a **R\$ 8,2 bilhões**, demonstrando o compromisso governamental com a área.

## Investimentos Financeiros

### Orçamento Total de Segurança Pública

**2024**: R\$ 7,8 bilhões[^1]

- Polícia Militar: R\$ 5,46 bilhões (70% do total)
- Polícia Civil: R\$ 2,1 bilhões
- Sistema Prisional: R\$ 2,6 bilhões

**2025**: R\$ 8,2 bilhões (estimativa)[^2][^3]

- Crescimento de 5,1% em relação a 2024
- Orçamento aprovado com déficit de R\$ 8,6 bilhões


### Recursos Federais

**2024**: R\$ 42,7 milhões[^4][^5]

- Recursos do Fundo Nacional de Segurança Pública (FNSP)
- Transferência antecipada em 3 meses comparado a 2023

**2025**: R\$ 43 milhões (estimativa)[^6]

- Governo federal empenhará R\$ 1,166 bilhão para todos os estados
- Minas deve receber proporção similar ao ano anterior


### Investimentos em Infraestrutura

**2024**: R\$ 112 milhões[^7]

- R\$ 27 milhões para reestruturação de 101 unidades da Polícia Civil
- R\$ 10,6 milhões para renovação da frota da PC
- R\$ 74,8 milhões para reforma de 12 unidades prisionais
- Entrega de 7.310 coletes balísticos

**2025**: Investimentos contínuos[^8]

- R\$ 11 milhões em equipamentos para Polícia Civil
- 32 drones, 51 viaturas, 587 pistolas, 114 fuzis para PC


## Principais Ações e Programas

### Modernização da Frota

**2024**:

- 165 viaturas entregues para Polícia Militar[^9][^10]
- 53 viaturas semi-blindadas (R\$ 16,5 milhões)[^9]
- 112 viaturas diversas (caminhonetes L-200 e Trailblazer)[^10]

**2025**:

- 274 viaturas entregues até julho[^8][^11]
- 223 veículos PM (201 motos + 22 carros)[^11]
- 51 viaturas para Polícia Civil[^8]
- Investimento total: R\$ 18,5 milhões[^11]


### Concursos Públicos e Fortalecimento do Efetivo

**2024**:

- Concurso Polícia Civil: 255 vagas[^12][^13]
- Concurso Polícia Militar: 3.102 vagas[^14]
- Formação de aproximadamente 3.000 novos policiais militares[^11]

**2025**:

- 359 nomeações previstas para Polícia Civil[^15][^16]
- Novo concurso PM: 2.500 vagas anunciadas[^17]
- Concurso administrativo PC: 104 vagas[^16]


### Operações Especiais

**2024**:

- Operação Carnaval 2024[^18]
- Operação Eleições 2024[^19]
- Operação Natalina 2024[^20][^21]
- Mega Operação Proclamação da República[^22]

**2025**:

- Operação Férias Seguras 2025[^23][^24]
- Operação Carnaval 2025[^25]
- Operação Semana Santa 2025[^26][^27]
- Operação Maio Amarelo 2025[^28]
- Operação PMMG 250 Anos[^29]


### Novos Benefícios

**2025**: Criação inédita de auxílio alimentação[^30][^31]

- R\$ 50/dia para servidores da segurança pública
- Previsão de expansão para R\$ 75/dia até final de 2025
- Impacto de até 23% no salário dos servidores da base
- Custo estimado: R\$ 220 milhões/ano


## Resultados Alcançados

### Indicadores de Segurança

**2024**:

- Minas Gerais reconhecido pelo IBGE como segundo estado mais seguro do Brasil[^18]
- Redução de 12,7% nos crimes violentos em 2023[^4]
- Queda de 54% na criminalidade violenta de 2019 a 2023[^32]

**2025**:

- Redução de 33% nos crimes violentos[^33][^34]
- Queda de 21,4% nos homicídios durante Carnaval 2025[^25]
- Redução de 7,48% nos crimes violentos no primeiro trimestre[^11]


### Tecnologia e Modernização

**2024**:

- Investimento de R\$ 5 milhões em sistemas de inteligência[^1]
- Metodologia de Integração da Gestão em Segurança Pública (Igesp)[^35]

**2025**:

- Entrega de equipamentos modernos: drones, armas de última geração[^8]
- Recebimento de equipamentos federais: 27.240 espargidores, 742 kits de armas eletrônicas[^36]


## Comparação e Análise

### Crescimento dos Investimentos

| Categoria | 2024 | 2025 | Variação |
| :-- | :-- | :-- | :-- |
| Orçamento Total | R\$ 7,8 bi | R\$ 8,2 bi | +5,1% |
| Recursos Federais | R\$ 42,7 mi | R\$ 43 mi | +0,7% |
| Equipamentos PC | R\$ 5 mi | R\$ 11 mi | +120% |

### Principais Diferenças

**2024** focou em:

- Estruturação da base: concursos massivos
- Modernização da frota policial
- Reformas de infraestrutura

**2025** priorizou:

- Valorização dos servidores (auxílio alimentação)
- Equipamentos tecnológicos avançados
- Operações especiais comemorativas (250 anos da PM)


## Desafios e Perspectivas

### Desafios Identificados

1. **Déficit Orçamentário**: O orçamento de 2025 prevê déficit de R\$ 8,6 bilhões[^2][^3]
2. **Déficit de Efetivo**: Polícia Civil opera com 35% de defasagem em relação ao ideal[^15][^16]
3. **Pressão Fiscal**: Gastos com pessoal ultrapassam limites da Lei de Responsabilidade Fiscal[^2]

### Perspectivas Futuras

1. **Concursos Contínuos**: Previsão de 2.500 vagas anuais para PM[^17]
2. **Modernização Tecnológica**: Investimentos em sistemas de inteligência e equipamentos
3. **Integração Regional**: Combate ao crime organizado nas divisas estaduais[^9]

## Conclusão

Minas Gerais demonstra consistência na ampliação dos investimentos em segurança pública, com crescimento de 5,1% no orçamento total entre 2024 e 2025. Os destaques incluem o aumento significativo (120%) nos equipamentos da Polícia Civil, a criação inédita do auxílio alimentação e o foco em operações especiais estratégicas.

O estado mantém sua liderança nacional em segurança pública, combinando investimentos estruturais com ações preventivas e repressivas coordenadas. Os resultados positivos nos indicadores de criminalidade validam a efetividade das estratégias adotadas, consolidando Minas Gerais como referência em gestão de segurança pública no Brasil.
