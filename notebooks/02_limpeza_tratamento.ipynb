# 🧼 Limpeza e Tratamento dos Dados de Crimes Violentos - MG


Este notebook documenta o processo de tratamento dos dados brutos de crimes violentos em Minas Gerais.

Etapas principais:
1. Leitura e unificação de múltiplos arquivos CSV.
2. Remoção de acentos, cedilha e caracteres especiais.
3. Criação de colunas úteis, como 'data'.
4. Salvamento de um dataset padronizado e limpo em CSV.


## 🧰 Scripts utilizados no tratamento

O tratamento de dados foi automatizado por meio dos seguintes scripts localizados na pasta `src/`:

- `uniao_dados.py`: Lê todos os arquivos CSV da pasta `data/raw`, aplica os tratamentos e gera o arquivo unificado `crimes_unificado_bruto.csv`.
- `limpeza.py`: Funções auxiliares para:
  - Normalização de nomes de colunas
  - Conversão de datas
  - Remoção de acentos e caracteres especiais
  - Padronização de tipos e limpeza geral

Esses scripts garantem que o processo seja reprodutível e padronizado.

## 🔄 1. Carregando dados unificados e tratados via script

```python
import pandas as pd

caminho_csv_tratado = '../data/processed/crimes_unificado_bruto.csv'
df = pd.read_csv(caminho_csv_tratado, sep=';', encoding='utf-8')
display(df.head())
```

## 📋 2. Estrutura esperada após tratamento

- Sem acentos ou cedilhas em nomes de colunas e valores
- Todas as colunas com nomes simples e minúsculos
- Coluna 'data' criada a partir de 'ano' e 'mes' no formato YYYY-MM-DD (dia fixo 01)
- Registros de todos os anos empilhados em um dataframe único

```python

display(df.info())

display(df.describe(include='all'))
```

## 🧪 3. Verificações pós-tratamento

```python

print("Colunas do dataframe:", list(df.columns))


print("Naturezas:", df['natureza'].unique())
print("RMBH:", df['rmbh'].unique())


print(f"Intervalo de datas: {df['data'].min()} -> {df['data'].max()}")
print("Amostra de datas:")
display(df['data'].sample(5, random_state=42))
```

## 🧹 4. Resumo das transformações aplicadas

- Leitura dos arquivos CSV da pasta `data/raw` (2012 a 2024)
- Unificação em um único dataframe
- Padronização dos nomes de colunas
- Conversão de 'ano' e 'mes' para inteiros
- Criação da coluna 'data' (datetime)
- Remoção de acentos e cedilha
- Exclusão de colunas 100% nulas ou irrelevantes (se houver)
- Salvamento em CSV no caminho: '../data/processed/crimes_unificado_bruto.csv'
