 ## 📌 Contexto do Projeto

O projeto foi estruturado em cinco etapas principais:

1. Exploração Inicial dos Dados  (01_exploracao_dados.ipynb)
   → Coleta, análise de estrutura e compreensão inicial dos dados.

2. Limpeza e Tratamento  (02_limpeza_tratamento.ipynb)

   → Criação do banco de dados SQLite, integração com dados populacionais do IBGE e normalização de chaves.

   → Geração de gráficos detalhados, modelagem preditiva com ARIMA, regressão linear e Random Forest, além da estimativa populacional e ajustes por 100 mil habitantes.

5. Conclusão Geral (este notebook)
   → Resumo crítico dos resultados.

## 🧠 Aprendizados Técnicos

-Gerenciamento eficiente de dados com SQLite: Estruturei e utilizei bancos de dados SQLite para armazenar, consultar e integrar grandes volumes de dados brutos e tratados, garantindo reuso, consistência e escalabilidade durante todas as fases da análise.

-Tratamento e limpeza avançada de dados: Desenvolvi rotinas para padronização textual, remoção de acentuação, normalização de nomes de municípios e integração de bases heterogêneas (dados de crimes e IBGE), o que foi fundamental para evitar inconsistências e duplicações.

-Modelagem preditiva com múltiplas técnicas: Implementei três abordagens de previsão — Regressão Linear, ARIMA e Random Forest — e realizei backtesting para validar e comparar seus desempenhos, garantindo uma escolha informada do melhor modelo para previsão de crimes em Minas Gerais.

-Análise de séries temporais e previsão com intervalos de confiança: Aprofundei o uso de modelos ARIMA para séries temporais e aprendi a interpretar e apresentar previsões junto com seus intervalos de confiança, evidenciando a incerteza inerente às previsões estatísticas.

-Visualizações interativas e informativas com Plotly e Matplotlib: Criei gráficos dinâmicos e visualmente atraentes, com anotações claras, legendas e uso de cores adequadas para facilitar a interpretação dos dados por diferentes públicos.

-Documentação e comunicação clara: Além da análise técnica, pratiquei a elaboração de narrativas coerentes para apresentar resultados, riscos e limitações, assegurando que os insights fossem acessíveis e contextualizados para tomadas de decisão.

## 📌 Contexto Complementar: Investimentos e Ações em Segurança Pública (2024–2025)

Como parte do encerramento deste projeto, é fundamental observar que os dados de criminalidade analisados estão inseridos em um ecossistema mais amplo de políticas públicas, investimentos e ações coordenadas pelo Estado. Abaixo, apresentamos uma síntese dos dados extraídos via Perplexity[^1], com foco nas iniciativas implementadas pelo Governo de Minas Gerais nos anos de 2024 e 2025.

---

### 💰 Evolução dos Investimentos

Minas Gerais ampliou o orçamento destinado à segurança pública em **5,1%**, passando de **R$ 7,8 bilhões em 2024** para **R$ 8,2 bilhões em 2025**. Essa ampliação é refletida em diversas frentes:

- **Infraestrutura**: Reestruturação de unidades policiais e prisionais.
- **Equipamentos**: Aquisição de viaturas, armamentos, coletes e drones.
- **Tecnologia**: Modernização dos sistemas de inteligência e operação.
- **Valorização Profissional**: Criação inédita de auxílio alimentação em 2025, com impacto de até 23% no salário de base dos servidores.

#### 📊 Comparativo de Destaques

| Indicador                             | 2024              | 2025              | Variação   |
|--------------------------------------|-------------------|-------------------|------------|
| Orçamento Total                      | R$ 7,8 bilhões     | R$ 8,2 bilhões     | +5,1%      |
| Recursos Federais                    | R$ 42,7 milhões    | R$ 43 milhões      | +0,7%      |
| Investimento em Equipamentos PC      | R$ 5 milhões       | R$ 11 milhões      | +120%      |

| Efetivo – Concursos                  | 3.357 vagas        | 2.963 vagas        | ↓          |

---

### 👮‍♂️ Ações Estratégicas

#### Modernização Operacional
- Aquisição de 32 drones, 51 viaturas, 587 pistolas e 114 fuzis só em 2025.
- Entrega de mais de 7.000 coletes balísticos em 2024.

#### Reforço no Efetivo
- Formação de cerca de 3.000 novos policiais militares em 2024.
- Nomeações e concursos previstos também em 2025, incluindo áreas administrativas da Polícia Civil.

#### Operações Especiais
- Execução de ações como Operação Carnaval, Natalina, Semana Santa, Férias Seguras e outras em ambos os anos.
- Operações comemorativas aos **250 anos da PMMG** em 2025.

---

### 📉 Resultados e Desafios

#### Indicadores Positivos
- Redução de **33%** nos crimes violentos em 2025.
- Queda de **21,4%** nos homicídios durante o Carnaval 2025.
- Estado reconhecido como o **2º mais seguro do Brasil** segundo o IBGE.

#### Desafios Relevantes
- **Déficit orçamentário** estadual previsto em R$ 8,6 bilhões para 2025.
- **Defasagem de efetivo**: Polícia Civil opera com apenas 65% da capacidade ideal.
- **Pressão fiscal** sobre os gastos com pessoal e infraestrutura.

# 📎 Considerações Finais

A criminalidade é um fenômeno multifatorial, influenciado por variáveis sociais, econômicas e institucionais. Ao integrar os dados orçamentários e operacionais da segurança pública ao panorama analítico, fica evidente que **ações de Estado** — como investimentos contínuos, concursos, aquisição de equipamentos e operações planejadas — podem ter **impacto direto e indireto** na redução dos crimes.

Essa perspectiva fortalece o entendimento de que análises preditivas e exploratórias de criminalidade ganham em profundidade quando contextualizadas com políticas públicas reais, sobretudo em um cenário estadual de desafios fiscais e busca por eficiência.
